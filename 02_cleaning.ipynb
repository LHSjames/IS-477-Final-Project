```python
import pandas as pd
import os

# Define the file path
DATA_DIR = 'data'
INTEGRATED_FILE = os.path.join(DATA_DIR, 'integrated_chicago_data.csv')

# Load the data
print(f"Loading {INTEGRATED_FILE}...")

df = pd.read_csv(INTEGRATED_FILE)
print("File loaded successfully!")

# See a summary of all columns, their types, and missing values
df.info()
```

```python
# Get a count of missing values for each column
missing_values = df.isnull().sum()
print("Missing Value Counts:")
print(missing_values[missing_values > 0])
```

```python
print(f"Columns before dropping: {df.shape[1]}")
print("Dropping 'neighbourhood_group' and 'neighbourhood_group_neighborhood'...")

df.drop(columns=['neighbourhood_group', 'neighbourhood_group_neighborhood'], inplace=True)

print("Empty columns dropped.")
print(f"Columns after dropping: {df.shape[1]}")

# See the updated list of columns
df.info()
```

```python
# Check how many rows have a missing price
print(f"Rows in dataframe (before): {len(df)}")
print(f"Rows with missing price: {df['price'].isnull().sum()}")

# Drop any rows where the price is missing (NaN)
df.dropna(subset=['price'], inplace=True)

print(f"Rows in dataframe (after dropping): {len(df)}")
print(f"Rows with missing price (after): {df['price'].isnull().sum()}")
```

```python
# Define a new output file
CLEANED_FILE = os.path.join(DATA_DIR, 'cleaned_chicago_data.csv')

print(f"Saving cleaned data to {CLEANED_FILE}...")
df.to_csv(CLEANED_FILE, index=False)
print(f"Done! Cleaned file saved to: {CLEANED_FILE}")
```
